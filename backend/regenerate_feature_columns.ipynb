# Regenerate feature_columns.pkl

This notebook will help you regenerate the `feature_columns.pkl` file in a format compatible with your current pandas version.

```python
import pandas as pd
import pickle
import joblib
import os

print(f"📦 Pandas version: {pd.__version__}")
print(f"📦 Python version: {os.sys.version}")
```

## Option 1: Inspect the existing pickle file

Try to extract column names from the corrupted file:

```python
# Try multiple methods to extract data
methods_tried = []
columns = None

# Method 1: Direct pickle load with latin1
try:
    with open('feature_columns.pkl', 'rb') as f:
        raw_data = pickle.load(f, encoding='latin1')
        print(f"Raw data type: {type(raw_data)}")
        print(f"Raw data: {raw_data}")
        
        # Try to extract as list
        if hasattr(raw_data, 'tolist'):
            columns = raw_data.tolist()
        elif hasattr(raw_data, 'values'):
            columns = list(raw_data.values)
        elif isinstance(raw_data, (list, tuple)):
            columns = list(raw_data)
        
        if columns:
            print(f"✅ Extracted {len(columns)} columns!")
            print(f"Columns: {columns}")
    methods_tried.append("pickle+latin1")
except Exception as e:
    print(f"❌ Method 1 failed: {e}")
    print(f"Error type: {type(e).__name__}")

# Method 2: Inspect pickle structure
if columns is None:
    print("\n Trying to inspect pickle structure...")
    try:
        import pickletools
        with open('feature_columns.pkl', 'rb') as f:
            pickletools.dis(f)
    except Exception as e:
        print(f"Could not analyze: {e}")
```

## Step 3: Save in compatible format

```python
if columns and len(columns) > 0:
    # Backup old file
    import shutil
    if os.path.exists('feature_columns.pkl'):
        shutil.copy('feature_columns.pkl', 'feature_columns_OLD.pkl')
        print("✅ Backed up old file to: feature_columns_OLD.pkl")
    
    # Save as simple list using joblib (most compatible)
    joblib.dump(columns, 'feature_columns.pkl')
    print("✅ Saved new feature_columns.pkl using joblib")
    
    # Also save as JSON for easy inspection
    import json
    with open('feature_columns.json', 'w') as f:
        json.dump(columns, f, indent=2)
    print("✅ Saved feature_columns.json for inspection")
    
    # Verify we can load it back
    loaded = joblib.load('feature_columns.pkl')
    print(f"\n✅ Verification: Loaded {len(loaded)} features")
    print(f"   First 5: {loaded[:5]}")
    
    print("\n🎉 SUCCESS! Your feature_columns.pkl is now compatible.")
    print("   Restart your backend server to use the new file.")
else:
    print("\n❌ No columns defined. Please fill in Option 2 with your actual features.")
```
